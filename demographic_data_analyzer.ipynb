```python
import pandas as pd

columns = [
    'age',
    'workclass',
    'fnlwgt',
    'education',
    'education-num',
    'marital-status',
    'occupation',
    'relationship',
    'race',
    'sex',
    'capital-gain',
    'capital-loss',
    'hours-per-week',
    'native-country',
    'salary'
]

df = pd.read_csv(
    'adult.data.csv',
    names=columns,
    skipinitialspace=True
)

print(df.head())
```

```python
country_salary = df.groupby('native-country')
```

```python
print(country_salary)
```

```python
country_percentages = (
    country_salary['salary']
    .apply(lambda x: (x == '>50K').mean() * 100)
)
```

```python
print(country_percentages)
```

```python
highest_earning_country = country_percentages.idxmax()
```

```python
print(highest_earning_country)
```

```python
highest_earning_country_percentage = round(
    country_percentages.max(),
    1
)
```

```python
print(highest_earning_country_percentage)
```
